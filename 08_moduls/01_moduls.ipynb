## Importacions de la biblioteca estàndard

`import math` permet accedir a `math.sqrt`. `from statistics import mean`
incorpora només el nom `mean`. Amb `as` podem donar un àlies local al mòdul.
Tots dos mòduls formen part de Python.

Prediu el resultat de les tres crides abans d'executar-les.
Utilitza importacions explícites per identificar l'origen de cada nom.

```python
import math
from statistics import mean
import math as matematiques

print(math.sqrt(25))  # 5.0
print(mean([20.0, 22.0, 24.0]))  # 22.0
print(matematiques.pi)  # 3.141592653589793
```

## Localitzar els mòduls propis al notebook

`conversions.py` i `sensors.py` són al costat d'aquest notebook.
Quan executem un script, Python cerca a la carpeta del script. En un notebook
la carpeta inicial depèn de com hem obert Jupyter o VS Code.

La cel·la següent prepara aquest entorn per a les dues ubicacions previstes,
l'arrel del repositori i la carpeta `08_moduls`. `Path.cwd()` consulta la
carpeta de treball i `sys.path` és la llista de carpetes on Python cerca mòduls.
Al bloc 10 aprofundirem en les rutes amb `Path`.

```python
import sys
from pathlib import Path

carpeta_moduls = Path.cwd()
if carpeta_moduls.name != "08_moduls":
    carpeta_moduls = carpeta_moduls / "08_moduls"
if str(carpeta_moduls) not in sys.path:
    sys.path.insert(0, str(carpeta_moduls))
```

## Reutilitzar dues responsabilitats

Obre [conversions.py](conversions.py) i [sensors.py](sensors.py).
El primer transforma unitats. El segon calcula una mitjana i retorna `None`
quan falten lectures. `None` expressa l'absència d'una mesura.

El programa que coordina aquestes funcions és el primer pas del projecte de
monitoratge. Cap sensor requereix maquinari real.

```python
import conversions as conv
from sensors import mitjana

print(conv.celsius_a_kelvin(25.0))  # 298.15
print(conv.milivolts_a_volts(3300))  # 3.3
print(mitjana([21.0, 22.0, 23.0]))  # 22.0
print(mitjana([]))  # None
```

## Coordinar funcions i protegir les demostracions

La funció següent utilitza els dos mòduls per presentar les lectures.
Separar càlcul i presentació facilita reutilitzar els càlculs en altres programes.

`__name__` val `"__main__"` en executar directament un fitxer. En importar-lo,
val el nom del mòdul. La guarda limita l'execució de la demostració al primer cas.
Als notebooks també acostuma a valer `"__main__"`.

```python
def mostrar_resum(lectures):
    """Mostra la temperatura mitjana simulada en Celsius i kelvins."""
    temperatura_mitjana = mitjana(lectures)
    if temperatura_mitjana is None:
        print("Encara no hi ha lectures.")
    else:
        print(f"Mitjana: {temperatura_mitjana:.1f} °C")
        print(f"Mitjana: {conv.celsius_a_kelvin(temperatura_mitjana):.2f} K")


if __name__ == "__main__":
    mostrar_resum([20.0, 21.0, 22.0])
    mostrar_resum([])
```

```python
# El nom d'un mòdul importat és el seu nom de fitxer sense l'extensió.
print(conv.__name__)  # conversions
# A conversions.py les demostracions estan dins de la guarda __main__.
# Per això la primera importació anterior no ha mostrat les seves proves.
```
