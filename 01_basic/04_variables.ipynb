# 04 — Variables

Les variables serveixen per desar dades a la memòria. Python és un llenguatge de tipatge dinàmic i fort.

## Assignar valors a variables

Per assignar un valor a una variable, escrivim el seu nom i fem servir `=`. Després podem consultar-ne el valor i el tipus amb `type()`.

````python
nom_complet = "marc esteve"
print(nom_complet)  # Mostra el valor de la variable nom_complet

edat = 32
print(edat)  # Mostra el valor de la variable edat
print(type(edat))  # Mostra el tipus de dada de la variable edat (int)
````

Es pot assignar un valor nou a una variable existent.

````python
edat = 39
print(edat)  # Ara la variable edat té el valor 39
````

## Tipatge dinàmic

El tipus de dada es determina durant l'execució. No cal declarar-lo explícitament i una variable pot rebre valors de tipus diferents.

````python
nom = "marc"
print(type(nom))  # Mostra el tipus de dada de la variable nom (str)

nom = 32
print(type(nom))  # Ara la variable conté un nombre enter (int)
````

## Tipatge fort

Python no converteix automàticament entre tipus incompatibles. Per exemple, sumar un enter i una cadena provoca un `TypeError`:

```python
print(10 + "2")
```

## Cadenes de text formatades (*f-strings*)

Una *f-string* permet inserir valors i expressions dins d'una cadena de text.

````python
print(f"Hola {nom_complet}, d'aquí a 5 anys tindré {edat + 5} anys")
````

El mateix missatge també es podria construir concatenant cadenes i convertint el resultat numèric amb `str()`:

```python
print("Hola " + nom_complet + ", d'aquí a 5 anys tindré " + str(edat + 5) + " anys")
```

## Convencions per als noms de variables

A Python se sol fer servir `snake_case`: paraules en minúscula separades per guions baixos. Els noms en majúscules se solen reservar per a constants, tot i que Python no les fa constants de manera efectiva.

````python
nom, edat, ciutat = "pepito", 32, "Madrid"

el_meu_nom_de_variable = "correcte"  # snake_case
nom = "correcte"

elMeuNomDeVariable = "no recomanat"  # camelCase
ElMeuNomDeVariable = "no recomanat"  # PascalCase
elmeunomdevariable = "no recomanat"  # tot junt

el_meu_nom_de_variable_123 = "correcte"

LA_MEVA_CONSTANT = 3.14  # UPPER_CASE -> constants
LA_MEVA_CONSTANT = 3  # Python no té constants, però se segueix aquesta convenció
````

Un nom de variable no pot començar amb un nombre ni contenir guions o espais. Tampoc no es poden fer servir paraules reservades de Python com a noms de variables. Aquests exemples provocarien errors i, per tant, no s'executen:

```python
123123_variable = "incorrecte"  # No pot començar amb un nombre
el-meu-nom = "incorrecte"  # No pot contenir guions (-); fes servir guions baixos (_)
el meu nom = "incorrecte"  # No pot contenir espais
True = False  # No es poden sobreescriure les paraules reservades
```

Entre les paraules reservades de Python hi ha `False`, `None`, `True`, `and`, `as`, `assert`, `async`, `await`, `break`, `class`, `continue`, `def`, `del`, `elif`, `else`, `except`, `finally`, `for`, `from`, `global`, `if`, `import`, `in`, `is`, `lambda`, `nonlocal`, `not`, `or`, `pass`, `raise`, `return`, `try`, `while`, `with` i `yield`.

## Anotacions de tipus

Les anotacions de tipus són opcionals i poden fer el codi més clar. No canvien el tipus del valor assignat.

````python
usuari_ha_iniciat_sessio: bool = True  # Indica que la variable és booleana
print(usuari_ha_iniciat_sessio)

nom: str = "marc"  # Indica que la variable és una cadena de text
print(nom)
````
