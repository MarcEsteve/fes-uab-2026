# 01 — `print()`

`print()` és la funció que permet mostrar informació a la consola. T'acompanyarà tota la vida: des d'avui fins a la fi dels temps.

## Mostrar text

Aquest és un exemple bàsic de com imprimir un text a la consola.

````python
print("Hola, món!")
````

També pots utilitzar cometes simples per imprimir text.

````python
print('Això també funciona amb una cometa')
````

## Imprimir diversos elements

Pots imprimir diversos elements separats per un espai.

````python
print("Python", "és", "genial")
````

El paràmetre `sep` permet definir com se separen els elements impresos.

````python
print("Python", "és", "brutal", sep = "-")
````

## Definir el final de la línia

El paràmetre `end` defineix què s'imprimeix al final de la línia. En aquest exemple, `end = "\n"` indica explícitament un salt de línia.

````python
print("Això s'imprimeix", end = "\n") # Aquí, 'end' té un salt de línia explícit
print("en una línia") # Això s'imprimeix a la línia següent
````

## Imprimir nombres

També es poden imprimir números directament.

````python
print(42)
````

També es poden imprimir expressions numèriques.

````python
print(3+2)
````

## Incloure cometes dobles dins d'un text

Si utilitzem cometes dobles dins d'una cadena delimitada per cometes dobles, es produeix un error de sintaxi. Aquest exemple es mostra com a text perquè no es pot executar:

```python
print("Això és una "polzada"")  # Això generaria un error de sintaxi
```

Hi ha diverses maneres correctes d'incloure les cometes.

### Solució 1: fer servir cometes simples

Delimita la cadena amb cometes simples per poder incloure-hi cometes dobles.

````python
print('Això és una "polzada" dins una cadena amb cometes simples')
````

### Solució 2: fer servir el caràcter d'escapament

El caràcter `\` permet incloure cometes dobles dins d'una cadena delimitada per cometes dobles.

````python
print("Això és una \"polzada\" dins d'una cadena amb cometes dobles")
````

### Solució 3: fer servir cometes triples

Les cometes triples també permeten definir la cadena.

````python
print("""Això és una "polzada" dins d'una cadena amb cometes triples""")
````
