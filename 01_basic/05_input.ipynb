# 05 — Entrada de dades amb `input()`

La funció `input()` permet obtenir dades de l'usuari a través de la consola. Rep un missatge que es mostra a l'usuari i retorna el valor que introdueix.

## Llegir una cadena de text

Executa la cel·la i escriu el teu nom quan se't demani.

```python
nom = input("Hola, com et dius?\n")
print(f"Hola {nom}, encantat de conèixer-te")
print("Hola " + nom + ", encantat de conèixer-te")
```

## Convertir l'entrada a nombre

Tingues en compte que `input()` retorna una cadena de text. Per fer operacions numèriques, cal convertir-la a un tipus numèric, com ara `int`.

```python
edat = input("Quants anys tens?\n")
print(f"D'aquí a 3 anys tindràs {int(edat) + 3} anys")
edat = int(edat)
print(f"Tens {edat} anys")
```

## Llegir diversos valors alhora

Per obtenir diversos valors amb una sola crida a `input()`, l'usuari els ha de separar amb espais. En aquest exemple, introdueix primer el país i després la ciutat.

```python
print("Obtenir diversos valors alhora")
pais, ciutat = input("A quin país i a quina ciutat vius?\n").split()
print(f"Vius a {ciutat}, {pais}")
```
