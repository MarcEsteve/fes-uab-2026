## Rutes i espai de treball

`Path` representa una ruta. L'operador `/` uneix noms de manera portable. Una ruta relativa es resol respecte del directori de treball.

`TemporaryDirectory` crea una carpeta de pràctica i la neteja quan acaba el bloc `with`. Cada exemple disposa de les seves dades temporals.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    ruta = Path(carpeta) / "temperatures.txt"
    print(ruta.name)
    print(ruta.exists())  # False, la ruta encara no té fitxer.
```

## Obrir, escriure i tancar

`open` retorna un fitxer. El mode `w` crea el fitxer o substitueix el contingut anterior. `encoding="utf-8"` conserva accents i símbols com `°C`. `with` garanteix el tancament del recurs.

El mode `r` llegeix. `read()` recupera tot el text. Abans d'executar, prediu si el salt de línia final també es recupera.

```python
with TemporaryDirectory() as carpeta:
    ruta = Path(carpeta) / "estat.txt"
    with open(ruta, "w", encoding="utf-8") as fitxer:
        fitxer.write("Sensor de temperatura actiu\n")
    with open(ruta, "r", encoding="utf-8") as fitxer:
        text = fitxer.read()
    print(repr(text))
```

## Una lectura per línia

Iterar sobre el fitxer permet processar les línies una a una. `strip()` elimina espais i salts de línia dels extrems. El fitxer conté text i `float` el transforma en dades numèriques.

Separem la persistència del càlcul perquè les funcions siguin reutilitzables. Un text invàlid genera `ValueError` i un fitxer absent genera `FileNotFoundError`.

```python
def desar_temperatures(ruta, temperatures):
    """Desa una temperatura per línia."""
    with open(ruta, "w", encoding="utf-8") as fitxer:
        for temperatura in temperatures:
            fitxer.write(f"{temperatura}\n")


def llegir_temperatures(ruta):
    """Recupera nombres i ignora línies buides."""
    temperatures = []
    with open(ruta, "r", encoding="utf-8") as fitxer:
        for linia in fitxer:
            text = linia.strip()
            if text:
                temperatures.append(float(text))
    return temperatures

with TemporaryDirectory() as carpeta:
    ruta = Path(carpeta) / "temperatures.txt"
    desar_temperatures(ruta, [20.0, 21.5])
    print(llegir_temperatures(ruta))  # [20.0, 21.5]
```

## Afegir lectures

El mode `a` afegeix contingut al final. Si falta el fitxer, el crea. Cada registre necessita un separador, aquí un salt de línia.

Prediu la diferència de resultat si substitueixes `a` per `w`.

```python
with TemporaryDirectory() as carpeta:
    ruta = Path(carpeta) / "temperatures.txt"
    desar_temperatures(ruta, [20.0, 21.5])
    with open(ruta, "a", encoding="utf-8") as fitxer:
        fitxer.write("22.0\n")
    print(llegir_temperatures(ruta))  # [20.0, 21.5, 22.0]
```

## Casos límit

Un fitxer existent sense lectures retorna `[]`. Un fitxer absent indica un error d'accés. Gestionem cada situació segons el contracte de la funció.

```python
with TemporaryDirectory() as carpeta:
    ruta = Path(carpeta) / "temperatures.txt"
    desar_temperatures(ruta, [])
    print(llegir_temperatures(ruta))  # []
    try:
        llegir_temperatures(Path(carpeta) / "absent.txt")
    except FileNotFoundError:
        print("El fitxer encara no existeix")
```
