# Explicació del primer programa en Python

Aquest notebook explica pas a pas el contingut del fitxer `exemple.py`.

## Codi original

El programa està format per una única línia de codi.

```python
print("Hola! Aquest es el meu primer programa en Python.")
```

## Explicació línia per línia

- `print` és una funció incorporada de Python que mostra informació per pantalla.
- Els parèntesis `()` indiquen les dades que passem a la funció.
- El text entre cometes és una cadena de caràcters, també anomenada *string*.
- Quan executem la cel·la, Python mostra aquesta cadena a la sortida del notebook.

```python
missatge = "Hola! Aquest es el meu primer programa en Python."
print(missatge)
```
