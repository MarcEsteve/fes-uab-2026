# 06 · Tuples

Col·leccions **ordenades** i **immutables** d'elements. A diferència de les llistes, **no es poden modificar** un cop creades.

## 1️⃣ Creació d'una tupla

````python
la_meva_tupla = (1, 2, 3, 2, 4, 2)
print("Tupla original:", la_meva_tupla)
````

## 2️⃣ Intentar modificar un valor (error)

```python
la_meva_tupla[0] = 5
# ❌ TypeError: 'tuple' object does not support item assignment
```

No ho executem perquè provocaria un error.

## 3️⃣ Solució: convertir a llista, modificar i tornar a tupla

````python
llista = list(la_meva_tupla)      # convertir a llista
llista[0] = 5                     # modificar el primer element
la_meva_tupla = tuple(llista)     # convertir de nou a tupla
print("Tupla modificada convertint-la en llista:", la_meva_tupla)
````

## 4️⃣ Crear una tupla nova amb valors modificats

````python
la_meva_tupla = (6,) + la_meva_tupla[1:] # Modifica el primer element de la tupla a 5
print("Tupla nova amb valors modificats:", la_meva_tupla)
````

## 5️⃣ Accedir als elements

````python
print("Primer element:", la_meva_tupla[0])
print("Últim element:", la_meva_tupla[-1])
````

## 6️⃣ Longitud

````python
print("Nombre d'elements de la tupla:", len(la_meva_tupla))
````

## 7️⃣ Mètodes disponibles

````python
print("Nombre de vegades que apareix el 2 a la tupla:", la_meva_tupla.count(2))
print("Índex de la primera aparició del 4:", la_meva_tupla.index(4))
````
