## Una classe descriu els seus objectes

`class Sensor` defineix el model. `__init__` inicialitza cada objecte nou. `self` fa referència a la instància que executa el mètode. `self.nom`, `self.unitat` i `self.lectures` són atributs. La llista es construeix dins d'`__init__` perquè cada sensor tingui la seva.

Un mètode és una funció definida a la classe. `registrar` comprova que el valor sigui un `int` o `float` finit abans d'afegir-lo. `bool` és un subtipus d'`int` i el rebutgem explícitament. `math.isfinite` identifica els floats infinits i `nan`.

Les consultes retornen `None` quan falta una lectura. La mitjana es calcula sumant les mesures i dividint pel recompte.

```python
from math import isfinite


class Sensor:
    """Agrupa la identificació i l'històric d'un sensor simulat."""

    def __init__(self, nom, unitat):
        """Inicialitza una instància amb un històric buit propi."""
        # Python crida __init__ quan construïm l'objecte.
        # self identifica la instància que està executant el mètode.
        self.nom = nom
        self.unitat = unitat
        # Aquesta llista es crea per a cada objecte, dins d'__init__.
        self.lectures = []

    # 2. MÈTODES I CANVIS D'ESTAT
    # En sensor.registrar(22.5), Python passa sensor com a self.
    # La validació es fa abans de modificar l'històric.
    def registrar(self, valor):
        """Afegeix un int o float finit, o llança ValueError."""
        # bool és un subtipus d'int, però True no és una mesura.
        if isinstance(valor, bool) or not isinstance(valor, (int, float)):
            raise ValueError("La lectura ha de ser un nombre.")
        # Els enters són finits. Els floats també poden ser inf o nan.
        if isinstance(valor, float) and not isfinite(valor):
            raise ValueError("La lectura ha de ser finita.")
        self.lectures.append(valor)

    # 3. MÈTODES QUE CONSULTEN L'ESTAT
    # None representa l'absència de lectura. 0 és una mesura vàlida.
    def ultima_lectura(self):
        """Retorna l'última lectura, o None si l'històric és buit."""
        if not self.lectures:
            return None
        return self.lectures[-1]

    def mitjana(self):
        """Retorna la mitjana aritmètica, o None si no hi ha lectures."""
        if not self.lectures:
            return None
        return sum(self.lectures) / len(self.lectures)
```

## Construir instàncies

Cada crida a `Sensor(...)` crea un objecte. Els arguments es passen a `__init__`. En accedir a un atribut fem servir el punt. Prediu els dos historials abans d'executar aquesta cel·la.

```python
temperatura = Sensor("T1", "°C")
humitat = Sensor("H1", "%")
print(temperatura.nom, temperatura.unitat)  # T1 °C
print(temperatura.lectures)  # []
print(humitat.lectures)  # []
```

## Canviar l'estat amb mètodes

En `temperatura.registrar(20.0)`, Python passa automàticament `temperatura` com a `self`. El mètode modifica l'històric d'aquesta instància. Dibuixa una caixa per a cada sensor i anota l'estat després de cada crida.

```python
temperatura.registrar(20.0)
temperatura.registrar(24.0)
humitat.registrar(55.0)
print(temperatura.lectures)  # [20.0, 24.0]
print(humitat.lectures)  # [55.0]
```

## Consultar l'estat

`ultima_lectura()` i `mitjana()` consulten l'històric. El valor `None` ens permet distingir l'absència de dades d'una mesura igual a zero. Aquesta cel·la inicia la segona sessió.

```python
print(temperatura.ultima_lectura())  # 24.0
print(temperatura.mitjana())  # 22.0
tensio = Sensor("V1", "V")
print(tensio.ultima_lectura())  # None
print(tensio.mitjana())  # None
tensio.registrar(0)
print(tensio.ultima_lectura())  # 0
print(tensio.mitjana())  # 0.0
```

## Validació abans del canvi

Reutilitzem `ValueError` i `try` del bloc 09. Si la lectura és invàlida, l'excepció arriba abans de `append`. Capturar-la permet continuar l'exemple. Comprova que es conserven les lectures vàlides.

```python
for valor_invalid in ["error", True, float("inf"), float("nan")]:
    try:
        temperatura.registrar(valor_invalid)
    except ValueError as error:
        print(error)
print(temperatura.lectures)  # [20.0, 24.0]
```

## Activitat guiada d'estat i instàncies

Prediu les últimes lectures després d'afegir un zero a T1. Explica per què H1 conserva el valor anterior. Després representa en un dibuix els tres objectes.

`Sensor` concentra la responsabilitat de gestionar les seves lectures. Centralitzar-ne la validació facilita la mantenibilitat. En aquest model l'atribut `lectures` és públic. Les mesures s'afegeixen amb `registrar` per conservar la validació.

```python
temperatura.registrar(0)
print(temperatura.ultima_lectura())  # 0
print(humitat.ultima_lectura())  # 55.0
```
