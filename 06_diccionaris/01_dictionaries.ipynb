# 05 · Diccionaris

Col·leccions de **parelles clau-valor**. Serveixen per emmagatzemar dades relacionades.

## 📦 Exemple típic de diccionari

```python
persona = {
    "nom": "marc",
    "edat": 25,
    "es_estudiant": True,
    "qualificacions": [7, 8, 9],
    "socials": {
        "twitterx": "@marcesteveg",
        "github": "@MarcEsteve",
        "linkedin": "Marc Esteve Garcia"
    }
}
```

## 🔑 Accedir als valors

```python
print(persona["nom"])  # "marc"
print(persona["edat"])  # 25
print(persona["qualificacions"][2])
print(persona["socials"]["twitterx"])
```

## ✏️ Canviar valors

```python
persona["nom"] = "esteve"
persona["qualificacions"][2] = 10
print(persona)
```

## 🗑️ Eliminar una propietat

Amb `del` o amb `pop()` (que a més retorna el valor).

```python
del persona["edat"]
print(persona)

es_estudiant = persona.pop("es_estudiant")
print(f"es_estudiant: {es_estudiant}")
print(persona)
```

## 🔄 Sobreescriure amb un altre diccionari (`update`)

```python
a = {"nom": "marc", "edat": 25}
b = {"nom": "daniel", "es_estudiant": False}
print(a)
a.update(b)
print(a)
```

## ❓ Comprovar si existeix una clau

```python
print("name" in persona)  # False
print("nom" in persona)   # True
```

## 🔍 `keys()`, `values()` i `items()`

```python
print("keys:")
print(persona.keys())
print(persona["socials"].keys())

print("\nvalues:")
print(persona.values())

print("\nitems:")
print(persona.items())
```

## 🔁 Recórrer un diccionari

```python
for clau, valor in persona.items():
    print(f"{clau}: {valor}")
```
