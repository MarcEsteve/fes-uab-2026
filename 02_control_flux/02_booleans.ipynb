# 02 — Valors booleans

Els valors lògics són `True` (cert) i `False` (fals). Són fonamentals per al control del flux i la lògica en programació.

## Valors booleans

Els booleans representen valors de veritat.

```python
print(True)
print(False)
```

## Operadors de comparació

Les comparacions retornen un valor booleà. Alguns operadors habituals són `>`, `<`, `==` (igual), `!=` (diferent), `>=` (major o igual) i `<=` (menor o igual).

```python
print("5 > 3:", 5 > 3)
print("5 < 3:", 5 < 3)
print("5 == 5:", 5 == 5)
print("5 != 3:", 5 != 3)
print("5 >= 5:", 5 >= 5)
print("5 <= 3:", 5 <= 3)
```

## Comparar cadenes de text

Les cadenes es poden comparar. La comparació d'igualtat distingeix entre majúscules i minúscules.

```python
print("'poma' < 'pera':", "poma" < "pera")
print("'Hola' == 'hola':", "Hola" == "hola")
```

## Operadors lògics

- `and` és cert si tots dos operands són certs.
- `or` és cert si almenys un dels operands és cert.
- `not` inverteix el valor booleà.

```python
print("True and True:", True and True)
print("True and False:", True and False)
print("True or False:", True or False)
print("False or False:", False or False)
print("not True:", not True)
print("not False:", not False)
```

## Taules de veritat

Les taules següents mostren els resultats de `and`, `or` i `not` per a cada combinació de valors booleans.

```python
print("and:")
print("A      B      A and B")
print("True   True ", True and True)
print("True   False", True and False)
print("False  True ", False and True)
print("False  False", False and False)

print("\nor:")
print("A      B      A or B")
print("True   True ", True or True)
print("True   False", True or False)
print("False  True ", False or True)
print("False  False", False or False)

print("\nnot:")
print("A      not A")
print("True  ", not True)
print("False ", not False)
```
